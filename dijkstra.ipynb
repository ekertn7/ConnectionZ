```python
graph = cnnnz.UndirectedGraph()
edges = [
    ('Andrew', 'Brandon', {'distance': 14}),
    ('Andrew', 'Brandon', {'distance': 6}),
    ('Brandon', 'Camila', {'distance': 18}),
    ('Brandon', 'Camila', {'distance': 11}),
    ('Camila', 'Delilah', {'distance': 19}),
    ('Brandon', 'Evelyn', {'distance': 7}),
    ('Evelyn', 'Evelyn', {'distance': 9}),
    ('Evelyn', 'Felix', {'distance': 3}),
    ('Gabriel', 'Felix', {'distance': 12}),
    ('Presley', 'Josiah', {'distance': 17})]
for node_l, node_r, attributes in edges:
    graph.add_edge(node_l, node_r, **attributes)
graph
```

```python
graph.get_node_attribute('Andrew', 'neighbors')
```

```python
for edge in graph.get_edges('Andrew', 'Brandon'):
    print(graph.get_edge_attribute('Andrew', 'Brandon', edge, 'distance'))
```
